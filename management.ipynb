```python
students = []

while True:
    print("\n===== Student Attendance Management System =====")
    print("1. Add Student")
    print("2. View Attendance")
    print("3. Calculate Attendance Percentage")
    print("4. Check Low Attendance")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        name = input("Enter student name: ")
        total_classes = int(input("Enter total classes: "))
        attended_classes = int(input("Enter classes attended: "))

        students.append({
            "name": name,
            "total": total_classes,
            "attended": attended_classes
        })

        print("Student added successfully!")

    elif choice == "2":
        if len(students) == 0:
            print("No student records found.")
        else:
            print("\n--- Student Attendance ---")

            for student in students:
                print("Name:", student["name"])
                print("Total Classes:", student["total"])
                print("Classes Attended:", student["attended"])
                print()

    elif choice == "3":
        if len(students) == 0:
            print("No student records found.")
        else:
            print("\n--- Attendance Percentage ---")

            for student in students:
                percentage = (
                    student["attended"] / student["total"]
                ) * 100

                print(
                    student["name"],
                    ":", 
                    round(percentage, 2),
                    "%"
                )

    elif choice == "4":
        if len(students) == 0:
            print("No student records found.")
        else:
            print("\n--- Low Attendance Students ---")

            for student in students:
                percentage = (
                    student["attended"] / student["total"]
                ) * 100

                if percentage < 75:
                    print(
                        student["name"],
                        "-",
                        round(percentage, 2),
                        "%"
                    )

    elif choice == "5":
        print("Thank you for using the system!")
        break

    else:
        print("Invalid choice. Please try again.")
```
